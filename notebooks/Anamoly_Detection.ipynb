Given sensor readings from turbofan engines over time, predict Remaining Useful Life (RUL) and detect early abnormal behavior before failure.
This is:
1. Regression (RUL prediction)
2. Time-series degradation modeling
3. Anomaly detection


---


*   Column 1: Corresponds to engine number (This column is indexed 0 above because of Python's numbering convention)
*   Column 2: Corresponds to cycle number. If engine 1 fails after 192 cycles, the entries of second column for engine 1 will go from 1 to 192. Similarly for other engines.
* Columns 3,4,5: 3 operational settings
* Columns 6-26: 21 sensor measurements

```python
"""
col_names = ['id','cycle'] + \
            [f'op_set_{i}' for i in range(1,4)] + \
            [f'sensor_{i}' for i in range(1,20)]
df = df.dropna(axis=1)
df.columns = col_names
"""
col_names = ['id','cycle'] + \
            [f'op_set_{i}' for i in range(1,4)] + \
            [f'sensor_{i}' for i in range(1,22)]
df = df.dropna(axis=1)
df.columns = col_names
```

```python
df.head(15)
```

```python
max_cycle = df.groupby('id')['cycle'].max().reset_index()
max_cycle.columns = ['id', 'max_cycle']

df = df.merge(max_cycle, on='id')
df['RUL'] = df['max_cycle'] - df['cycle']
df=df.drop(['max_cycle'], axis=1)
df
```

```python
engine_1 = df[df['id'] == 1]
engine_1
```

```python
sns.set(style="whitegrid")

# Select engine
id = 1
engine_df = df[df['id'] == id]

# Get sensor columns dynamically
sensor_cols = [col for col in df.columns if "sensor_" in col]

# Create subplots grid (7 rows × 3 columns = 21 sensors)
fig, axes = plt.subplots(nrows=7, ncols=3, figsize=(18, 20), sharex=True)

axes = axes.flatten()

for i, sensor in enumerate(sensor_cols):
    axes[i].plot(engine_df['cycle'], engine_df[sensor])
    axes[i].set_title(sensor)
    axes[i].set_ylabel("Value")

plt.tight_layout()
plt.show()
```

```python
plt.figure()
plt.plot(engine_1['cycle'], engine_1['RUL'])
plt.xlabel("Cycle")
plt.ylabel("Remaining Useful Life (RUL)")
plt.title(f"Engine {id} - RUL Over Time")
plt.gca().invert_yaxis()
plt.show()
```

```python
# Get only sensor columns
sensor_cols = [col for col in df.columns if "sensor_" in col]

# Compute variance
variance = df[sensor_cols].var().sort_values()

print(variance)
```

```python
low_variance_sensors = variance[variance < 1e-5].index.tolist()
print("Low variance sensors:", low_variance_sensors)

df.drop(columns=low_variance_sensors, inplace=True)
```

```python
df.shape
```

```python
corr_with_rul = df.corr()['RUL'].sort_values()
print(corr_with_rul)
```

```python
important_sensors = corr_with_rul.abs().sort_values(ascending=False).head(8).index.tolist()
important_sensors.remove('RUL')

print("Important sensors:", important_sensors)
important_sensors = [s for s in important_sensors if s != 'cycle']
print("Important sensors:", important_sensors)
```

```python
window = 5

for sensor in important_sensors:
    df[f"{sensor}_rolling_mean"] = df.groupby('id')[sensor].transform(lambda x: x.rolling(window, min_periods=1).mean())
    df[f"{sensor}_rolling_std"] = df.groupby('id')[sensor].transform(lambda x: x.rolling(window, min_periods=1).std())
```

```python
# Cap RUL
RUL_CAP = 130
df['RUL'] = df['RUL'].clip(upper=RUL_CAP)

print("RUL capped at:", RUL_CAP)
print(df['RUL'].describe())
```

**Why 130?** :-
Because very high RUL values behave like flat healthy region and confuse regression.

```python
from sklearn.model_selection import train_test_split

# Get unique engine IDs
ids = df['id'].unique()
# Split engines
train_ids, test_ids = train_test_split(
    ids,
    test_size=0.2,
    random_state=42
)

# Create train & test datasets
train_df = df[df['id'].isin(train_ids)]
test_df  = df[df['id'].isin(test_ids)].copy()

print("Train Engines:", len(train_ids))
print("Test Engines:", len(test_ids))
```

```python
# Columns to remove
drop_cols = ['id', 'cycle', 'RUL']

# If max_cycle exists
if 'max_cycle' in df.columns:
    drop_cols.append('max_cycle')

# Feature columns
feature_cols = [col for col in df.columns if col not in drop_cols]

print("Total Features:", len(feature_cols))
print(feature_cols)
```

```python
X_train = train_df[feature_cols]
y_train = train_df['RUL']

X_test = test_df[feature_cols]
y_test = test_df['RUL']
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled  = scaler.transform(X_test)
```

```python
print("Train shape:", X_train_scaled.shape)
print("Test shape:", X_test_scaled.shape)
print("Target shape:", y_train.shape)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error

# Fill NaN values in scaled data (e.g., with 0 for rolling std at min_periods=1)
X_train_scaled = np.nan_to_num(X_train_scaled, nan=0.0)
X_test_scaled = np.nan_to_num(X_test_scaled, nan=0.0)

# Train
lr = LinearRegression()
lr.fit(X_train_scaled, y_train)

# Predict
y_pred_lr = lr.predict(X_test_scaled)

# Metrics
rmse_lr = np.sqrt(mean_squared_error(y_test, y_pred_lr))
mae_lr = mean_absolute_error(y_test, y_pred_lr)

print("Linear Regression Results")
print("RMSE:", rmse_lr)
print("MAE :", mae_lr)
```

```python
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(
    n_estimators=200,
    max_depth=None,
    random_state=42,
    n_jobs=-1
)

rf.fit(X_train, y_train)

y_pred_rf = rf.predict(X_test)

rmse_rf = np.sqrt(mean_squared_error(y_test, y_pred_rf))
mae_rf = mean_absolute_error(y_test, y_pred_rf)

print("\nRandom Forest Results")
print("RMSE:", rmse_rf)
print("MAE :", mae_rf)
```

```python
print("\nModel Comparison")
print(f"Linear Regression RMSE: {rmse_lr:.2f}")
print(f"Random Forest RMSE   : {rmse_rf:.2f}")
print(f"Linear Regression MAE: {mae_lr:.2f}")
print(f"Random Forest MAE    : {mae_rf:.2f}")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))
plt.scatter(y_test, y_pred_rf, alpha=0.3)
plt.xlabel("Actual RUL")
plt.ylabel("Predicted RUL")
plt.title("Random Forest: Actual vs Predicted RUL")
plt.plot([0,125],[0,125], color='red')  # diagonal line
plt.show()
```

```python
feature_importance = pd.Series(
    rf.feature_importances_,
    index=feature_cols
).sort_values(ascending=False)

top_10 = feature_importance.head(10)

print("\nTop 10 Important Features:")
print(top_10)
```

```python
plt.figure(figsize=(8,6))
top_10.sort_values().plot(kind='barh')
plt.title("Top 10 Feature Importance (Random Forest)")
plt.show()
```

```python
from sklearn.ensemble import IsolationForest

iso = IsolationForest(
    n_estimators=200,
    contamination=0.05,   # assume 5% anomalies
    random_state=42,
    n_jobs=-1
)
healthy_df = df[df['RUL'] > 100]
iso.fit(healthy_df[feature_cols])
```

```python
# Anomaly score (higher = more normal, lower = more abnormal)
df['anomaly_score'] = iso.decision_function(df[feature_cols])

# Flag (-1 = anomaly, 1 = normal)
df['anomaly_flag'] = iso.predict(df[feature_cols])

print(df[['anomaly_score', 'anomaly_flag']].head())
```
